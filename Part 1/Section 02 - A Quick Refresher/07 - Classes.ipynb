```python
class Rectangle:
    def __init__(self, width, height) -> None:
        self.width = width
        self.height = height
        
        
    # we can use any keyword instead of self.
    # def area(obj):
    #     return obj.height * obj.width
    
    # def perimeter(obj):
    #     return 2 * (obj.width + obj.height)
    
    
    def area(self):
        return self.height * self.width
    
    def perimeter(self):
        return 2 * (self.width + self.height)
    
    
    def __str__(self):
        return "With print and str() for Human."
    
    
    def __repr__(self):
        return "Without print for Developer"
    
    
    
    
r = Rectangle(10,20)
print("Area :" ,r.area())
print("Perimeter :" ,r.perimeter())
print(r)



        
```
