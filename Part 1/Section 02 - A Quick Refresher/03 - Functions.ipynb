#### User defined Functions

```python
def fun(name):
    print(f"Hi ,{name}")
fun
```

```python
def multiply(a: int , # here :int is defined for documentation. Nothing to wory with interpreter.
                b: int):   
    return a * b
```

```python
print(multiply(10,20))
```

```python
print(multiply(10.0,20))  # if we pass float then also executed.
```

```python
print(multiply('a',10))
```

```python
print(multiply([10,20,30],3))
```

```python
print(multiply([10,20,30],[3]))
```
