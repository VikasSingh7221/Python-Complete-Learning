```python
# While with Else Block.

data = [10, 20, 30, 40]
val = 100
found = False

idx = 0
while idx < len(data):
    if data[idx] == val:
        found = True
        break
    idx += 1

if not found:
    print("Not Found !!! So appending ...")
    data.append(val)
print(data)
    
```

```python
data = [10, 20, 30, 40, 100]
val = 100

idx = 0
while idx < len(data):
    if data[idx] == val:
        break
    idx += 1
else:
    print("Not Found !!! So appending ...")
    data.append(val)
    
print(data)
```

```python
a = 2
b = -2

while b < 2:
    print("========================")
    
    a -= 1
    b += 1
    
    try:
        a / b # pyright: ignore[reportUnusedExpression]
        
    except ZeroDivisionError:
        print("{0} , {1} inside Exception ".format(a, b))
        continue
    finally:
        print("{0} , {1} Always Executes ".format(a, b))
        
    print("{0} , {1} Main Loop ".format(a, b))
```
